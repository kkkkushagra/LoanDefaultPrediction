```python
from pathlib import Path
import sys
import pandas as pd
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
sys.path.insert(0, str(root))
from src.data.loading import load_application_data

path = root / "data" / "application_train.csv"
if not path.exists():
    raise FileNotFoundError(path)
df = load_application_data(path)
print(f"rows={len(df):,}, columns={df.shape[1]}")
```

```python
balance = df["TARGET"].value_counts().rename("count").to_frame()
balance["share"] = balance["count"] / len(df)
balance
```

```python
missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing.head(20).rename("missing_percent")
```
